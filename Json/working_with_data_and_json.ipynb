Working with JSON Data

JSON (JavaScript Object Notation) is a lightweight data-interchange format.
It's easy for humans to read and write and easy for
machines to parse. In Python, JSON maps almost perfectly to dictionaries.

json.dump(data, file_object): Writes a Python dictionary to a JSON file.
json.dumps(data): Converts a Python dictionary to a JSON formatted string.
json.load(file_object): Reads a JSON file into a Python dictionary.
json.loads(string): Parses a JSON formatted string into a Python dictionary.


```python
import json
data = {"name":"Alice","age":30}
json_string=json.dumps(data)
print(type(json_string))
print(json_string)
```

```python
import json 
student_grades = {
    'Alice':88,
    'Bob':92,
    'Charile':78,
    'students':[
        {'id':101,'name':'Alice','marks':55},
        {'id':102,'name':'Bob','marks':65}
    ]
}
with open('grades.json','w')as file:
    json.dump(student_grades,file)

print("grades.json created.")
```

```python
with open('grades.json','r')as file:
    data=json.load(file)

count=0
sum=0
for d in data['students']:
    count+=1
    sum+=d['marks']

average_marks=sum/count
print(average_marks)
```

```python
import json
with open('grades.json','r')as file:
    data=json.load(file)

print("Data loaded from JSON file")
print(type(data))
print(data)
print(f"Bob's grade is {data['Bob']}")
```

```python
try:
    with open('report.csv', 'r') as file:
        content = file.read()
    print(content)
except FileNotFoundError:
    print("Error: The file was not found.")
except PermissionError:
    print("Error: You do not have permission to read this file.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
else:
    print("Else block is executed")
finally:
    print("File operation finished (or failed).")
```
