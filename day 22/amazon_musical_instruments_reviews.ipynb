```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib
import seaborn as sns
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import accuracy_score, classification_report,confusion_matrix
from sklearn.linear_model import Perceptron
```

```python
df = pd.read_csv("amazon_musical_instruments_reviews.csv")
```

```python
df.head()
```

```python
df['overall'].value_counts
```

```python
df.isna().sum()
```

```python
df=df.dropna(subset=['reviewerName'])
```

```python
df=df.dropna(subset=['reviewText'])
```

```python
df.isna().sum()
```

```python
x=df['reviewText']
y=df['overall']


vectorizer=CountVectorizer()
x=vectorizer.fit_transform(x)

joblib.dump(vectorizer,"amazon_musical_instruments_reviews.pkl")
```

```python
from imblearn.over_sampling import SMOTE
smote=SMOTE()
x,y=smote.fit_resample(x,y)
```

```python
y.value_counts()
```

```python
X_train, X_test, y_train, y_test = train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
model=Perceptron()
model.fit(X_train,y_train)

y_pred=model.predict(X_test)
print(y_pred)   
joblib.dump(model,"model_amazon_musical_instruments_reviews.pkl")

```

```python

print(confusion_matrix(y_test,y_pred))

#Detailed classification report 
print("\nClassification Report: ")
print(classification_report(y_test,y_pred))
```
