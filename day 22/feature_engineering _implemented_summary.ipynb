```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib
import seaborn as sns
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import accuracy_score, classification_report,confusion_matrix
from sklearn.linear_model import Perceptron
```

```python
df = pd.read_csv("amazon_musical_instruments_reviews.csv")
```

```python
df=pd.DataFrame(df)
```

```python
df.head()
```

```python
df=df.dropna(subset=['reviewerName'])
df=df.dropna(subset=['reviewText'])
```

```python
def f(row):
   
    if row['overall'] == 3.0:
        val = 'Neutral'

    elif row['overall'] == 1.0 or row['overall'] == 2.0:
        val = 'Negative'

    elif row['overall'] == 4.0 or row['overall'] == 5.0:
        val = 'Positive'

    else:
        val = -1

    return val

df['reviews'] = df.apply(f, axis=1)
df.head()
```

```python
df['reviews']=df['summary']+' '+df['reviewText']
```

```python
x=df['reviewText']
y=df['overall']


vectorizer=CountVectorizer()
x=vectorizer.fit_transform(x)

joblib.dump(vectorizer,"new_amazon_musical_instruments_reviews.pkl")
```

```python
from imblearn.over_sampling import SMOTE
smote=SMOTE()
x,y=smote.fit_resample(x,y)
```

```python
y.value_counts()
```

```python
X_train, X_test, y_train, y_test = train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
model=MLPClassifier(hidden_layer_sizes=(100,),solver="lbfgs")
model.fit(X_train,y_train)

y_pred=model.predict(X_test)
print(y_pred)   
joblib.dump(model,"new_model_amazon_musical_instruments_reviews.pkl")

```

```python

print(confusion_matrix(y_test,y_pred))

#Detailed classification report 
print("\nClassification Report: ")
print(classification_report(y_test,y_pred))
```
